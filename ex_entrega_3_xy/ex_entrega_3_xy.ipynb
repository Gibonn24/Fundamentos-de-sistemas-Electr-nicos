# Desarrollo

Funções de Ativação, Treinamento e Inferência

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

def derivada_sigmoid(y):
    return y * (1 - y)

def treinamento(x, d, eta, Ne, Nh, Nb=20):
    """
    Treinamento da rede neural (Generalização da Aula 4).
    x: sinal de entrada (Nt, Ni)
    d: sinal desejado (Nt, 1)
    eta: passo de adaptação (learning rate)
    Ne: número de épocas
    Nh: número de neurônios na camada oculta
    Nb: tamanho do mini-batch
    """
    Nt, Ni=x.shape
    Nmb=int(np.floor(Nt/Nb))

    #W1: pesos da camada oculta, dimensão (Ni + 1, Nh)
    W1=0.5*np.random.randn(Ni+1,Nh)
    #W2: pesos da camada de saída, dimensão (Nh + 1, 1)
    W2=0.5*np.random.randn(Nh+1,1)

    eta=eta/Nb
    J_MSE=np.zeros(Ne)
    Xd=np.hstack((np.ones((Nt, 1)), x, d))

    for k in range(Ne):
        np.random.shuffle(Xd)
        X_shuf=Xd[:,0:Ni+1] #entrada
        D_shuf=Xd[:,[Ni+1]] #saida

        for l in range(Nmb):
            X_batch = X_shuf[l*Nb:(l+1)*Nb,:]
            D_batch = D_shuf[l*Nb:(l+1)*Nb,:]

            #forward
            V1=X_batch @ W1
            Y1=sigmoid(V1)

            #Adicionando bias para a camada de saída
            Y1_bias = np.hstack((np.ones((Nb, 1)), Y1))
            V2=Y1_bias @ W2
            Y2=sigmoid(V2)

            #erro
            E = D_batch - Y2

            #backwar
            #Gradientes locais
            delta2=E*derivada_sigmoid(Y2)
            #Ignoramos a primeira linha de W2 na retropropagação pois corresponde ao bias
            delta1=(delta2 @ W2[1:, :].T)*derivada_sigmoid(Y1)

            #Atualização dos pesos
            W2+=eta*(Y1_bias.T @ delta2)
            W1+=eta*(X_batch.T @ delta1)

            #Acúmulo do erro
            J_MSE[k] += np.sum(E ** 2)

        #Calcula o MSE da época
        J_MSE[k] = J_MSE[k] / (Nmb * Nb)

    return W1, W2, J_MSE

def inferencia(x, W1, W2):
    """
    Realiza a predição da rede utilizando os pesos treinados.
    """
    Nt = x.shape[0]
    X_bias = np.hstack((np.ones((Nt, 1)), x))
    V1 = X_bias @ W1
    Y1 = sigmoid(V1)
    Y1_bias = np.hstack((np.ones((Nt, 1)), Y1))
    V2 = Y1_bias @ W2
    Y2 = sigmoid(V2)

    return Y2
```

Geração dos Dados e Configurações Globais

```python
np.random.seed(42)

N_treino=5000
N_teste=1000

datasets={}

#a) f(x) = 1/x, x in [1, 100]
x_a_tr=np.random.uniform(1, 100, (N_treino, 1)); d_a_tr = 1 / x_a_tr
x_a_te=np.sort(np.random.uniform(1, 100, (N_teste, 1)), axis=0); d_a_te = 1 / x_a_te
datasets['a'] = (x_a_tr, d_a_tr, x_a_te, d_a_te, "f(x) = 1/x")

#b)f(x) = log10(x), x in [1, 10]
x_b_tr=np.random.uniform(1, 10, (N_treino, 1)); d_b_tr = np.log10(x_b_tr)
x_b_te=np.sort(np.random.uniform(1, 10, (N_teste, 1)), axis=0); d_b_te = np.log10(x_b_te)
datasets['b'] = (x_b_tr, d_b_tr, x_b_te, d_b_te, "f(x) = log10(x)")

#c)f(x)=exp(-x), x in [1, 10]
x_c_tr=np.random.uniform(1, 10, (N_treino, 1)); d_c_tr = np.exp(-x_c_tr)
x_c_te=np.sort(np.random.uniform(1, 10, (N_teste, 1)), axis=0); d_c_te = np.exp(-x_c_te)
datasets['c'] = (x_c_tr, d_c_tr, x_c_te, d_c_te, "f(x) = exp(-x)")

#d)f(x)=sen(x), x in [0, pi/2]
x_d_tr=np.random.uniform(0, np.pi/2, (N_treino, 1)); d_d_tr = np.sin(x_d_tr)
x_d_te=np.sort(np.random.uniform(0, np.pi/2, (N_teste, 1)), axis=0); d_d_te = np.sin(x_d_te)
datasets['d'] = (x_d_tr, d_d_tr, x_d_te, d_d_te, "f(x) = sen(x)")

#e)f(x,y)=sen(x)cos(y), (x,y) in [-2pi, 2pi] x [-2pi, 2pi]
x1_e_tr=np.random.uniform(-2*np.pi, 2*np.pi, (N_treino, 1))
x2_e_tr=np.random.uniform(-2*np.pi, 2*np.pi, (N_treino, 1))
x_e_tr=np.hstack((x1_e_tr, x2_e_tr))
d_e_tr=np.sin(x1_e_tr) * np.cos(x2_e_tr)
#Normalização para a Função E
d_e_tr_norm=(d_e_tr+1)/2

x1_e_te=np.random.uniform(-2*np.pi, 2*np.pi, (N_teste, 1))
x2_e_te=np.random.uniform(-2*np.pi, 2*np.pi, (N_teste, 1))
x_e_te=np.hstack((x1_e_te, x2_e_te))
d_e_te=np.sin(x1_e_te) * np.cos(x2_e_te)

datasets['e']=(x_e_tr, d_e_tr_norm, x_e_te, d_e_te, "f(x,y) = sen(x)cos(y)")
```

treinamento

```python
neuronios_ocultos=[3, 4, 5, 10, 15, 20, 50, 100]
epocas=2000
eta=0.1

tabela_mse = []

for func_key, (x_tr, d_tr, x_te, d_te, nome_func) in datasets.items():
    print(f"--- Treinando para: {nome_func} ---")

    if func_key != 'e':
        fig, axes = plt.subplots(2, 4, figsize=(20, 10))
        fig.suptitle(f"Aproximações para {nome_func}", fontsize=16)
        axes = axes.flatten()
    else:
        fig_x0, axes_x0 = plt.subplots(2, 4, figsize=(20, 10))
        fig_x0.suptitle("Função E com X=0, variando Y", fontsize=16)
        axes_x0 = axes_x0.flatten()

        fig_y0, axes_y0 = plt.subplots(2, 4, figsize=(20, 10))
        fig_y0.suptitle("Função E com Y=0, variando X", fontsize=16)
        axes_y0 = axes_y0.flatten()

    for i, Nh in enumerate(neuronios_ocultos):
        #Treinamento
        W1, W2, _ = treinamento(x_tr, d_tr, eta=eta, Ne=epocas, Nh=Nh, Nb=32)

        #Teste/Inferência
        y_pred=inferencia(x_te, W1, W2)
        if func_key == 'e':
            y_pred = (y_pred * 2) - 1

        #Cálculo do MSE
        mse=np.mean((d_te - y_pred)**2)
        tabela_mse.append({'Função': func_key.upper(), 'Neurônios (Nh)': Nh, 'MSE Teste': mse})


        if func_key != 'e':
            ax = axes[i]
            ax.scatter(x_te, d_te, color='blue', s=10, label="Alvo (Real)")
            ax.scatter(x_te, y_pred, color='red', s=10, label="Rede (Predito)")
            ax.set_title(f"Nh = {Nh} (MSE: {mse:.4f})")
            ax.legend()
        else:
            #Para a função E, criamos novos dados específicos para o gráfico:
            #1. x=0 fixo, variando y
            y_var = np.sort(np.random.uniform(-2*np.pi, 2*np.pi, (500, 1)), axis=0)
            x_fixo = np.zeros_like(y_var)
            entrada_x0 = np.hstack((x_fixo, y_var))
            alvo_x0 = np.sin(x_fixo) * np.cos(y_var)
            pred_x0 = (inferencia(entrada_x0, W1, W2) * 2) - 1

            axes_x0[i].plot(y_var, alvo_x0, 'b-', label="Alvo")
            axes_x0[i].plot(y_var, pred_x0, 'r--', label="Predito")
            axes_x0[i].set_title(f"Nh = {Nh}")

            #2. y=0 fixo, variando x
            x_var = np.sort(np.random.uniform(-2*np.pi, 2*np.pi, (500, 1)), axis=0)
            y_fixo = np.zeros_like(x_var)
            entrada_y0 = np.hstack((x_var, y_fixo))
            alvo_y0 = np.sin(x_var) * np.cos(y_fixo)
            pred_y0 = (inferencia(entrada_y0, W1, W2) * 2) - 1

            axes_y0[i].plot(x_var, alvo_y0, 'b-', label="Alvo")
            axes_y0[i].plot(x_var, pred_y0, 'r--', label="Predito")
            axes_y0[i].set_title(f"Nh = {Nh}")

    plt.tight_layout()
    plt.show()

# Exibindo a Tabela de MSE Final
df_mse = pd.DataFrame(tabela_mse)
df_pivot = df_mse.pivot(index='Função', columns='Neurônios (Nh)', values='MSE Teste')
print("\n=== Tabela de Erro Quadrático Médio (MSE) no Conjunto de Teste ===")
display(df_pivot)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Configuramos un número adecuado de neuronas y épocas para la función 3D
Nh_3d = 50
epocas_3d = 2000
eta_3d = 0.1

print(f"Entrenando red con Nh={Nh_3d} para la visualización 3D...")


x_e_tr, d_e_tr_norm, _, _, _ = datasets['e']
W1_3d, W2_3d, _ = treinamento(x_e_tr, d_e_tr_norm, eta=eta_3d, Ne=epocas_3d, Nh=Nh_3d, Nb=32)
rango = np.linspace(-2*np.pi, 2*np.pi, 100)
X_mesh, Y_mesh = np.meshgrid(rango, rango)
X_flat = X_mesh.flatten().reshape(-1, 1)
Y_flat = Y_mesh.flatten().reshape(-1, 1)
entradas_3d = np.hstack((X_flat, Y_flat))
Z_real = np.sin(X_mesh) * np.cos(Y_mesh)
Z_pred_flat = inferencia(entradas_3d, W1_3d, W2_3d)
Z_pred_flat = (Z_pred_flat * 2) - 1
Z_pred = Z_pred_flat.reshape(X_mesh.shape)

fig = plt.figure(figsize=(16, 7))
fig.suptitle(f"Comparaçao 3D: f(x,y) = sen(x)cos(y) | Neuronas Ocultas: {Nh_3d}", fontsize=16)

#real
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf1 = ax1.plot_surface(X_mesh, Y_mesh, Z_real, cmap='viridis', edgecolor='none')
ax1.set_title("Superficie Original (Alvo)")
ax1.set_xlabel("Eje X")
ax1.set_ylabel("Eje Y")
ax1.set_zlabel("Eje Z")
fig.colorbar(surf1, ax=ax1, shrink=0.5, aspect=10)

#entrenado
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
surf2 = ax2.plot_surface(X_mesh, Y_mesh, Z_pred, cmap='plasma', edgecolor='none')
ax2.set_title("Aproximación de la Red Neuronal")
ax2.set_xlabel("Eje X")
ax2.set_ylabel("Eje Y")
ax2.set_zlabel("Eje Z")
fig.colorbar(surf2, ax=ax2, shrink=0.5, aspect=10)

plt.tight_layout()
plt.show()
```
